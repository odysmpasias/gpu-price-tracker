```python
prices_per_card = df.groupby("id")["price"].nunique()
prices_per_card.value_counts()
```

## How often do prices change?

Number of distinct prices per card over 11 days (2026-10-21 to 2026-11-01, 24/10 missing):

- **10 cards** had two different prices.
- **No card** had more than two different prices.

Only about 7% of the 135 cards changed price. Prices on Plaisio are very stable and move in single "steps" (e.g. the MSI RTX 3050 went from 269 to 299 on 2026-10-28) rather than small daily changes.

**What this means for forecasting:** with this little data, "same price as yesterday" would be right ~93% of the time. More weeks of data are needed, and a more useful question may be *when* and *in which direction* a price changes.

```python
print(prices_per_card)
```

```python
changed_cards = prices_per_card[prices_per_card > 1]
changed_cards
```

```python
changed_rows = df[df["id"].isin(changed_cards.index)]
changed_rows.head(15)
```

```python
changed_rows = changed_rows.sort_values(["id","date"])
```

```python
price_change = changed_rows.groupby(["id", "name"])["price"].agg(["first", "last"])
```

```python
price_change
```

```python
price_change["diff"] = price_change["last"] - price_change["first"]
```

```python
price_change
```

```python
price_change["pct"] = ((price_change["diff"]/price_change["first"])*100).round(1)
```

```python
price_change
```

## Which cards changed price, and by how much?

All 10 cards with two different prices, comparing their first and last price (`diff` in €, `pct` in % of the first price):

- **9 of 10 got more expensive.** Only the PNY RTX 6000 Ada (a pro card) dropped, by 1.2%.
- **All 10 are NVIDIA.**
- **The RTX 5070 series moved the most:** +12.5% to +50%. Both Asus RTX 5070 cards rose by almost half.
- **The RTX 3050 cards rose less:** +3% to +16%.
- **The RTX 5090 rose only 2.9%.** Its +200€ looks large but is small relative to its price, which is why we compare percentages and not euros.
