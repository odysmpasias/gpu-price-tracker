## How often do prices change?

- **10 cards** had two different prices.

## Why do versions of the same model cost so differently?

Example: **RTX 5060 Ti** on the latest day (6 versions, 699–999€).

| VRAM | Price |
|---|---|
| 8 GB | 699–749€ |
| 16 GB | 999€ |

- **VRAM explains the big gap:** the 16 GB versions cost about 300€ (~40%) more than the 8 GB ones.
- **The product line explains the small gap:** among the 8 GB cards, the MSI GAMING OC (749€) is a higher-tier line than the VENTUS (699€).
- All three 16 GB versions cost exactly 999€, regardless of brand or line.

**What this means for forecasting:** "RTX 5060 Ti" is really two products. Grouping by `model` + `vram` will give more meaningful prices than `model` alone.

## Does the product line affect price? (as of 2026-11-13)

**Method:** For each card, `price_ratio = price / model_median`, where `model_median` is the median price of all versions of the same model on the latest day. A ratio of 1.10 means the card costs 10% more than a typical version of its model. This lets us compare lines fairly across cheap and expensive models. We removed cards that are the only version of their model (`model_count == 1`), since they have nothing to be compared with.

**Findings:**
- **VENTUS** (MSI) is the cheapest line, about 6–10% below the model median. This held on both days we checked (12/11 and 13/11) and is based on 8–11 cards.
- **ROG** (Asus) is not a premium line in this store: its cards sit at the model median. On 12/11 it looked slightly cheaper (0.97), but that was a one-day change.
- **AORUS, TUF and NITRO** look more expensive than average, but each has only 2–3 cards, so this is a hint, not a result.
- Most other lines are at or very close to 1.0, i.e. no clear difference.

**Limits:**
- Only one day of data per comparison and few cards per line, so the medians can move when a single card is added, removed or repriced.
- Prices come from one store (Plaisio) and may not represent the whole market.

**Side check:** The 9,999€ prices of the RTX 5090 cards are real store prices, not placeholders (checked on the site on 2026-11-13).

**Question:** Around 11–13/11 the median price index dropped by about 12%. Was this a store-wide price drop, or did it affect only some cards?

**Method:** We compare each card's price on **2026-11-09** (the last calm day before the drop) with its price on **2026-11-16**. Only gaming cards (`is_pro == False`) listed on both days are included. The dates are fixed, so the results below do not change when new data is added.

**Findings:**
- The drop is **not** store-wide. About 30% of the cards got cheaper, most kept the same price and a few got more expensive.
- The earlier median chart was misleading because it only included cards whose price changed at some point (selection bias). Cards with a stable price were left out.
- The cheaper cards are mostly **AMD**: AMD cards are about a quarter of the store but about half of the price cuts. Roughly 2 in 3 AMD cards got cheaper, compared to about 1 in 5 NVIDIA cards.
- The new prices are round retail prices (e.g. 399€, 499€, 999€), which looks like a planned promotion rather than random changes.

**Limits:**
- We cannot tell **why** AMD cards dropped: a store promotion (e.g. Black Friday) or a price cut by AMD itself. Data from one store cannot separate the two.
- The numbers are small (about 17 AMD cards), so this is a strong hint, not proof.

```python
all_prices = df[df["is_pro"]== False].pivot(index="date", columns="id", values="price")
```

```python
before = all_prices.loc["2026-11-09"]
after = all_prices.loc["2026-11-16"]
```
